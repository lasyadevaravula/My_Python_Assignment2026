```python
# 17. Which feature is most positively correlated with MPG?
import matplotlib.pyplot as plt
import seaborn as sns
correlation_matrix = df.corr()
correlation_with_mpg = correlation_matrix['mpg']
most_positively_correlated_feature = correlation_with_mpg.idxmax()
plt.figure(figsize=(8, 6))
sns.barplot(x=correlation_with_mpg.index, y=correlation_with_mpg.values)
plt.xlabel('Features')
plt.ylabel('Correlation with MPG')
plt.title('Correlation with MPG')
plt.xticks(rotation=45)
plt.show()
```

```python
# 18. Which feature is most negatively correlated with MPG?
import matplotlib.pyplot as plt
import seaborn as sns
correlation_matrix = df.corr()
correlation_with_mpg = correlation_matrix['mpg']
most_negatively_correlated_feature = correlation_with_mpg.idxmin()
plt.figure(figsize=(8, 6))
sns.barplot(x=correlation_with_mpg.index, y=correlation_with_mpg.values)
plt.xlabel('Features')
plt.ylabel('Correlation with MPG')
plt.title('Correlation with MPG')
plt.xticks(rotation=45)
plt.show()
```

```python
# 24. Detect outliers in MPG using IQR.
Q1 = df['mpg'].quantile(0.25)
Q3 = df['mpg'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df['mpg'] < lower_bound) | (df['mpg'] > upper_bound)]
print("Outliers in MPG:", outliers)
```
